### ۱.۴ مدیریت خطا و Decorator

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError as e:
        print(f"⚠️ خطا: {e}")
        return None
    except TypeError as e:
        print(f"⚠️ نوع داده اشتباه: {e}")
        return None
    finally:
        print("   (بلاک finally همیشه اجرا می‌شود)")

safe_divide(10, 0)

# Decorator: اندازه‌گیری زمان اجرای تابع
import time

def timer(func):
    def wrapper(*args, **kwargs):
        t0 = time.time()
        out = func(*args, **kwargs)
        print(f"⏱ {func.__name__} در {time.time()-t0:.4f} ثانیه اجرا شد")
        return out
    return wrapper

@timer
def heavy_sum(n):
    return sum(i*i for i in range(n))

print(heavy_sum(1_000_000))
````

### ۱.۵ Generator و Iterator

````python
def read_in_batches(n, batch=100):          # Generator با yield: حافظه‌کارآمد
    for i in range(0, n, batch):
        yield list(range(i, min(i + batch, n)))

gen = read_in_batches(250, batch=100)
print(next(gen))   # هر بار فقط یک بچ برمی‌گرداند
print(next(gen))
print("باقی‌مانده:", list(gen))

big_sum = sum(x*x for x in range(1_000_000) if x % 3 == 0)   # Generator Expression
print("جمع مربع‌های مضرب ۳:", big_sum)
````

### ۱.۶ شیءگرایی (OOP) پیشرفته: کلاس، ارث‌بری، Magic Method

````python
class Course:
    def __init__(self, name, points):
        self.name = name
        self.points = points
        self._students = []          # attribute «خصوصی» با قرارداد _

    def enroll(self, student):
        self._students.append(student)
        return self                 # زنجیره‌سازی متدها

    @property
    def student_count(self):
        return len(self._students)

    def __len__(self):              # Magic Method
        return len(self._students)

    def __repr__(self):
        return f"Course({self.name!r}, {self.points} Poäng)"

class OnlineCourse(Course):          # ارث‌بری
    def __init__(self, name, points, platform):
        super().__init__(name, points)
        self.platform = platform

    def enroll(self, student):       # Override متد
        print(f"ثبت‌نام آنلاین {student} در {self.platform}")
        return super().enroll(student)

c = OnlineCourse("Machine Learning", 30, "Canvas")
c.enroll("Sara").enroll("Ali")
print(c, "| تعداد دانشجو:", len(c))
````

### ۲.۴ جبر خطی (Linear Algebra)

````python
A = np.array([[2., 1.], [1., 3.]])
B = np.array([[5., 6.], [7., 8.]])

print("ضرب درایه‌ای:\n", A * B)
print("ضرب ماتریسی:\n", A @ B)                     # یا np.dot(A, B)
print("دترمینان:", np.linalg.det(A))
print("معکوس:\n", np.linalg.inv(A))
print("رده ماتریس (rank):", np.linalg.matrix_rank(A))

eigvals, eigvecs = np.linalg.eig(A)
print("مقادیر ویژه:", eigvals)
print("بردارهای ویژه:\n", eigvecs)

# حل دستگاه معادلات خطی: A x = b  →  x = A⁻¹b
b = np.array([5., 10.])
x = np.linalg.solve(A, b)                            # پایدارتر از inv(A) @ b
print("حل دستگاه:", x, "| بررسی:", A @ x)
````

````python
df = pd.DataFrame({
    "order_id": range(1001, 1021),
    "date": pd.date_range("2026-01-01", periods=20, freq="D"),
    "city": rng.choice(["Stockholm", "Göteborg", "Malmö", "Uppsala"], 20),
    "category": rng.choice(["Electronics", "Clothing", "Sports", "Furniture"], 20),
    "sales": np.round(rng.uniform(100, 3000, 20), 1),
    "quantity": rng.integers(1, 6, 20).astype(float),
    "rating": np.round(rng.uniform(1, 5, 20), 1),
})
# تزریق داده گم‌شده و تکراری برای تمرین واقعی
df.loc[rng.choice(df.index, 5, replace=False), "sales"] = np.nan
df.loc[rng.choice(df.index, 4, replace=False), "rating"] = np.nan
df.loc[rng.choice(df.index, 3, replace=False), "city"] = None
df = pd.concat([df, df.iloc[[0, 1]]], ignore_index=True)   # ردیف تکراری

print("=== shape:", df.shape)
df.info()
display(df.head(8))
print("\nآمار توصیفی:\n", df.describe().round(2))
print("\nتعداد مقادیر گم‌شده در هر ستون:\n", df.isna().sum())
````

````python
print(df.loc[0:4, ["city", "sales"]])                 # برچسب‌محور
print(df.iloc[0:5, 0:3])                              # موقعیت‌محور
print(df.query("sales > 2000 and category == 'Electronics'")[["order_id", "sales", "rating"]])

# دسته‌بندی فروش با cut
df["sales_band"] = pd.cut(df["sales"], bins=[0, 500, 1500, 3000], labels=["Low", "Mid", "High"])
print(df["sales_band"].value_counts())
````

### ۳.۳ پاک‌سازی کامل: تکراری‌ها و گم‌شده‌ها (جایگزینی گروهی)

````python
print("قبل:", df.shape)
df = df.drop_duplicates().reset_index(drop=True)       # حذف ردیف تکراری
print("بعد از حذف تکراری:", df.shape)

df = df.dropna(subset=["order_id", "sales"])           # حذف ردیف‌های بدون شناسه/فروش

# جایگزینی هوشمند: میانه/میانگین هر گروه — الگوی حرفه‌ای با transform
df["sales"] = df["sales"].fillna(df.groupby("category")["sales"].transform("median"))
df["rating"] = df["rating"].fillna(df.groupby("category")["rating"].transform("mean"))
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["quantity"] = df["quantity"].fillna(1)

print("\nگم‌شده باقی‌مانده:\n", df.isna().sum())
display(df.head())
````

### ۳.۴ Feature Engineering: apply، np.where و رشته‌ها

````python
df["revenue"] = df["sales"] * df["quantity"]
df["is_high_value"] = np.where(df["revenue"] > 4000, "Yes", "No")

def rating_label(r):                     # apply با تابع سفارشی
    if r >= 4: return "Excellent"
    if r >= 3: return "Good"
    return "Poor"

df["rating_label"] = df["rating"].apply(rating_label)

df["city_upper"] = df["city"].str.upper()
df["city_clean"] = df["city"].str.strip().str.title()
print(df[["city", "city_upper", "rating_label"]].head())
````

### ۳.۵ گروه‌بندی چندسطحی، transform و Pivot

````python
# agg چندتایی روی چند ستون
summary = (df.groupby(["city", "category"])
             .agg(total_revenue=("revenue", "sum"),
                  avg_rating=("rating", "mean"),
                  n_orders=("order_id", "count"),
                  max_sale=("sales", "max"))
             .reset_index())
display(summary)

# transform: «میانگین شهر» برای هر ردیف — بدون کوچک‌کردن جدول (معادل Window Function در SQL)
df["city_avg_revenue"] = df.groupby("city")["revenue"].transform("mean")
df["above_city_avg"] = df["revenue"] > df["city_avg_revenue"]

# Pivot Table و Crosstab
pivot = pd.pivot_table(df, values="revenue", index="city", columns="category",
                       aggfunc="sum", fill_value=0, margins=True, margins_name="Total")
print("=== Pivot Table ==="); display(pivot)
print("=== Crosstab (تعداد سفارش) ===")
display(pd.crosstab(df["city"], df["sales_band"]))
````

## ۴. SQL در پایتون — sqlite3 (Join، Window Function، CTE)

تمام مفاهیم درس SQL (۲۵ امتیاز) را مستقیماً در نوت‌بوک اجرا می‌کنیم.

````python
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

# ساخت دو جدول: کارمندان و فروش
cur.executescript('''
CREATE TABLE employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT,
    salary REAL
);
CREATE TABLE sales_sql (
    sale_id INTEGER PRIMARY KEY,
    emp_id INTEGER REFERENCES employees(emp_id),
    amount REAL,
    sale_date TEXT
);
''')

employees = [(1,"Sara","Data",52000),(2,"Ali","Data",48000),(3,"Erik","IT",55000),
             (4,"Mina","IT",61000),(5,"Jonas","Sales",45000)]
sales_sql = [(1,1,1200,"2026-01-05"),(2,1,900,"2026-01-12"),(3,2,1500,"2026-01-07"),
             (4,3,800,"2026-01-15"),(5,4,2000,"2026-01-20"),(6,5,600,"2026-01-09"),
             (7,2,1100,"2026-01-25"),(8,None,300,"2026-01-30")]   # فروش بدون کارمند (برای OUTER JOIN)

cur.executemany("INSERT INTO employees VALUES (?,?,?,?)", employees)
cur.executemany("INSERT INTO sales_sql VALUES (?,?,?,?)", sales_sql)
conn.commit()
print("✅ دیتابیس آماده شد")
````

### ۴.۱ JOINها و GROUP BY / HAVING

````python
q = """SELECT e.department,
              COUNT(*) AS n_employees,
              ROUND(AVG(e.salary), 0) AS avg_salary,
              SUM(s.amount) AS total_sales
       FROM employees e
       JOIN sales_sql s ON e.emp_id = s.emp_id
       GROUP BY e.department
       HAVING total_sales > 1000
       ORDER BY total_sales DESC;"""
display(pd.read_sql_query(q, conn))

print("INNER JOIN:", len(pd.read_sql_query(
    "SELECT * FROM employees e JOIN sales_sql s ON e.emp_id=s.emp_id", conn)))
print("LEFT JOIN:", len(pd.read_sql_query(
    "SELECT * FROM employees e LEFT JOIN sales_sql s ON e.emp_id=s.emp_id", conn)))
````

### ۴.۲ Window Functions: ROW_NUMBER، RANK، LAG/LEAD

````python
q = """SELECT name, department, salary,
       ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS rn,
       RANK()       OVER (PARTITION BY department ORDER BY salary DESC) AS rnk,
       AVG(salary)  OVER (PARTITION BY department) AS dept_avg,
       LAG(salary)  OVER (PARTITION BY department ORDER BY salary) AS prev_salary
       FROM employees;"""
display(pd.read_sql_query(q, conn))

# مجموع تجمیعی (running total) فروش هر کارمند
q2 = """SELECT emp_id, sale_date, amount,
        SUM(amount) OVER (PARTITION BY emp_id ORDER BY sale_date) AS running_total
        FROM sales_sql ORDER BY emp_id, sale_date;"""
display(pd.read_sql_query(q2, conn))
````

### ۴.۳ CTE و Subquery

````python
q = """WITH dept_stats AS (
           SELECT department, AVG(salary) AS avg_sal
           FROM employees GROUP BY department
       )
       SELECT e.name, e.salary, d.avg_sal,
              e.salary - d.avg_sal AS diff_from_avg
       FROM employees e
       JOIN dept_stats d ON e.department = d.department
       ORDER BY diff_from_avg DESC;"""
display(pd.read_sql_query(q, conn))

# Subquery در WHERE: کارمندانی بالاتر از میانگین کل شرکت
q2 = """SELECT name, salary FROM employees
        WHERE salary > (SELECT AVG(salary) FROM employees);"""
display(pd.read_sql_query(q2, conn))
conn.close()
````

## ۷. یادگیری عمیق با PyTorch
ساخت MLP، حلقه آموزش کامل، ارزیابی و تحلیل Loss.

````python
# آماده‌سازی Tensorها
scaler_dl = StandardScaler()
Xtr = torch.tensor(scaler_dl.fit_transform(X_train.values), dtype=torch.float32)
ytr = torch.tensor(y_train.values, dtype=torch.long)
Xte = torch.tensor(scaler_dl.transform(X_test.values), dtype=torch.float32)
yte = torch.tensor(y_test.values, dtype=torch.long)

class WineMLP(nn.Module):
    def __init__(self, in_dim, hidden1=32, hidden2=16, n_classes=3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, hidden1), nn.ReLU(),
            nn.Linear(hidden1, hidden2), nn.ReLU(),
            nn.Linear(hidden2, n_classes),
        )
    def forward(self, x):
        return self.net(x)

model = WineMLP(Xtr.shape[1])
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

losses = []
for epoch in range(200):                    # حلقه آموزش کامل
    optimizer.zero_grad()
    loss = criterion(model(Xtr), ytr)
    loss.backward()
    optimizer.step()
    losses.append(loss.item())

plt.plot(losses); plt.title("منحنی Loss در طول آموزش")
plt.xlabel("Epoch"); plt.ylabel("CrossEntropy"); plt.show()

model.eval()
with torch.no_grad():
    test_pred = model(Xte).argmax(dim=1)
    acc = (test_pred == yte).float().mean().item()
print(f"✅ دقت شبکه عصبی روی تست: {acc:.4f}")
print(classification_report(yte.numpy(), test_pred.numpy(), target_names=wine.target_names))
````

## ۸. پایتون پیشرفته: Data Pipeline شیءگرا
الگوی حرفه‌ای برای پروژه‌های LIA و Capstone.

````python
class DataPipeline:
    '''خط‌لوله شیءگرا: بارگذاری → پاک‌سازی → ویژگی → آماده‌سازی برای مدل'''
    def __init__(self, target_col):
        self.target_col = target_col
        self.df = None
        self.X, self.y = None, None

    def load(self, data):                       # قبول DataFrame یا مسیر CSV
        self.df = pd.read_csv(data) if isinstance(data, (str, Path)) else data.copy()
        return self

    def clean(self):
        df = self.df.drop_duplicates()
        num_cols = df.select_dtypes(include=[np.number]).columns
        df[num_cols] = df[num_cols].fillna(df[num_cols].median())
        cat_cols = df.select_dtypes(include=["object"]).columns
        for c in cat_cols:
            df[c] = df[c].fillna(df[c].mode()[0])
        self.df = df
        return self

    def build_features(self, drop_cols=None):
        df = self.df.drop(columns=drop_cols or [], errors="ignore")
        self.X = df.drop(columns=[self.target_col])
        self.y = df[self.target_col]
        return self

    def split(self, test_size=0.2):
        return train_test_split(self.X, self.y, test_size=test_size,
                                random_state=RANDOM_STATE, stratify=self.y)

# تست پایپ‌لاین روی داده شراب
wine_df = wine.frame
pipe = DataPipeline(target_col="target").load(wine_df).clean().build_features()
Xtr2, Xte2, ytr2, yte2 = pipe.split()
print("Pipeline: X_train", Xtr2.shape, "| X_test", Xte2.shape)
````

### ۱۰.۱ توزیع‌های احتمال با SciPy

````python
from scipy import stats

# توزیع نرمال: محاسبه PDF و CDF
mu, sigma = 50, 10
print("P(X < 60) در نرمال(50,10):", stats.norm.cdf(60, mu, sigma).round(4))
print("کمیت ۹۵٪ (percentile):", stats.norm.ppf(0.95, mu, sigma).round(2))

# توزیع دوجمله‌ای و پواسون
print("P(X=3) در دوجمله‌ای(n=10, p=0.3):", stats.binom.pmf(3, 10, 0.3).round(4))
print("P(X=2) در پواسون(λ=4):", stats.poisson.pmf(2, 4).round(4))

# نمونه‌گیری از توزیع‌ها و رسم
fig, axes = plt.subplots(1, 3, figsize=(15, 3.5))
x = np.linspace(20, 80, 300)
axes[0].plot(x, stats.norm.pdf(x, mu, sigma)); axes[0].set_title("Normal(50, 10)")
axes[1].bar(range(11), stats.binom.pmf(range(11), 10, 0.3)); axes[1].set_title("Binomial(10, 0.3)")
axes[2].bar(range(15), stats.poisson.pmf(range(15), 4)); axes[2].set_title("Poisson(4)")
plt.tight_layout(); plt.show()
````

### ۱۰.۲ قضیه بیز — مثال کاربردی (فیلتر اسپم)

````python
# P(Spam | کلمه "gratis") با فرمول بیز
p_spam = 0.2                       # فرض: ۲۰٪ ایمیل‌ها اسپم‌اند
p_gratis_given_spam = 0.7          # P(gratis | spam)
p_gratis = 0.25                    # P(gratis) در کل ایمیل‌ها

p_spam_given_gratis = (p_gratis_given_spam * p_spam) / p_gratis
print(f"P(Spam | gratis) = {p_spam_given_gratis:.2%}")

# Naive Bayes واقعی با scikit-learn روی داده متنی
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

texts = ["gratis vinnare erbjudande", "möte schema projekt", "gratis pengar nu",
         "rapport deadline leverans", "gratis bonus klicka här"]
labels = [1, 0, 1, 0, 1]           # 1 = spam

vec = CountVectorizer()
X_txt = vec.fit_transform(texts)
nb = MultinomialNB().fit(X_txt, labels)
print("پیش‌بینی برای 'gratis möte':", nb.predict(vec.transform(["gratis möte"])))
````

### ۱۰.۳ آمار استنباطی: آزمون فرض و p-value

````python
group_a = rng.normal(100, 15, 50)     # گروه کنترل
group_b = rng.normal(108, 15, 50)     # گروه تیمار

t_stat, p_value = stats.ttest_ind(group_a, group_b)
print(f"t-statistic: {t_stat:.3f} | p-value: {p_value:.5f}")
print("تفاوت معنادار است (α=0.05):", p_value < 0.05)

# فاصله اطمینان ۹۵٪ برای میانگین گروه A
ci = stats.t.interval(0.95, len(group_a)-1, loc=group_a.mean(), scale=stats.sem(group_a))
print("فاصله اطمینان ۹۵٪ میانگین A:", np.round(ci, 2))
````
